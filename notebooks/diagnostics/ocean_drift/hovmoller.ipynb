```python
from aqua.diagnostics.ocean_drift.hovmoller import Hovmoller
from aqua.diagnostics.ocean_drift.plot_hovmoller import PlotHovmoller

%matplotlib inline
```

```python
hov = Hovmoller(
    catalog="climatedt-gen2",
    model="IFS-NEMO-5km",
    exp="baseline-hist",
    source="lra-r100-monthly",
    startdate="01-01-1991",
    enddate="31-05-1991",
    # vert_coord="depth",
    loglevel="INFO",
)
```

```python
hov.run(region="io", dim_mean=["lat", "lon"], anomaly_ref="t0")
```

```python
hov_plot = PlotHovmoller(data=hov.processed_data_list, vert_coord="depth")
hov_plot.plot_hovmoller()
```
