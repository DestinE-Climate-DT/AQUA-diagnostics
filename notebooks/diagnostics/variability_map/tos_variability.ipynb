## Variability map diagnostic for Sea Surface Temperature

```python
%matplotlib inline
from aqua.diagnostics import VariabilityMap, PlotVariabilityMap
```

```python
dataset_dict = {"catalog": "climatedt-o25.1", "model": "IFS-FESOM", "exp":"historical-1990", "source":"lra-r100-monthly"}
dataset_dict_ref = {"catalog": "obs", "model": "ESA-CCI-L4", "exp":"v3.0.1", "source":"r100-monthly"}
```

```python
startdate="1994-01-01"
enddate="1994-03-31"
startdate_ref="1994-01-01"
enddate_ref="1994-03-31"
```

```python
# Intialize the TOS compute class
tos_dataset = VariabilityMap(
                    **dataset_dict,
                    var='tos',
                    startdate=startdate,
                    enddate=enddate,)
```

```python
# Run the compute function and save as NetCDF
tos_dataset.run()
```

```python
# Intialize the tos compute class for Reference data ESA-CCI-L4
tos_dataset_ref = VariabilityMap(
                    **dataset_dict_ref,
                    var='tos',
                    startdate=startdate,
                    enddate=enddate,)
```

```python
# Run the compute function and save as NetCDF
tos_dataset_ref.run()
```

```python
# Initialize the TOS plot class
plot_class = PlotVariabilityMap()
```

```python
# It is important to choose a correct 'tgt_grid_name' so that the data aligns with land mask
tgt_grid_name = "r1440x721"
# tgt_grid_name = "r1440x720"
```

```python
# Plot TOS for model dataset
# It is important to choose a correct 'tgt_grid_name' so that the data aligns with land mask
plot_dataset = {"catalog": "climatedt-o25.1", "model": "IFS-FESOM", "exp":"historical-1990", "tgt_grid_name":tgt_grid_name}
plot_class.plot(dataset_std=tos_dataset.data_std, **plot_dataset, startdate=startdate, enddate=enddate,)
```

```python
# Plot TOS for reference dataset
# It is important to choose a correct 'tgt_grid_name' so that the data aligns with land mask
plot_dataset_ref = {"catalog": "obs", "model": "ESA-CCI-L4", "exp":"v3.0.1", "tgt_grid_name":tgt_grid_name,}
plot_class.plot(dataset_std=tos_dataset_ref.data_std, **plot_dataset_ref, startdate=startdate, enddate=enddate,)
```

```python
# Plot the sub region for model data
# It is important to choose a correct 'tgt_grid_name' so that the data aligns with land mask
sub_region_dataset = {"catalog": "climatedt-o25.1", "model": "IFS-FESOM", "exp":"historical-1990", 'region': 'Agulhas', 'lon_limits': [5, 50], 'lat_limits': [-10, -50], 'proj': 'plate_carree', 'proj_params': {}, "tgt_grid_name":tgt_grid_name,}
plot_class.plot(dataset_std=tos_dataset.data_std, **sub_region_dataset, startdate=startdate, enddate=enddate,)
```

```python
# Plot the sub region for Reference data
# It is important to choose a correct 'tgt_grid_name' so that the data aligns with land mask
sub_region_dataset_ref = {"catalog": "obs", "model": "ESA-CCI-L4", "exp":"v3.0.1", 'region': 'Agulhas', 'lon_limits': [5, 50], 'lat_limits': [-10, -50], 'proj': 'plate_carree', 'proj_params': {}, "tgt_grid_name":tgt_grid_name,}
plot_class.plot(dataset_std=tos_dataset_ref.data_std, **sub_region_dataset_ref, startdate=startdate, enddate=enddate,)
```

```python
# Plot the diference of sub region for model dataset and reference dataset
time_intervals = {"startdate": "1994-01-01", "enddate": "1994-03-31", "startdate_ref": "1994-01-01", "enddate_ref": "1994-03-31"}
region_selection = {'region': 'Agulhas', 'lon_limits': [5, 50], 'lat_limits': [-10, -50], 'proj': 'plate_carree', 'proj_params': {},}
_dataset_ref = {"catalog_ref": "obs", "model_ref": "ESA-CCI-L4", "exp_ref":"v3.0.1",}
_dataset = {"catalog": "climatedt-o25.1", "model": "IFS-FESOM", "exp":"historical-1990",}
plot_class.plot_diff(dataset_std=tos_dataset.data_std, dataset_std_ref=tos_dataset_ref.data_std, tgt_grid_name=tgt_grid_name, **_dataset, **_dataset_ref, **region_selection, **time_intervals)
```
