# Ocean heat budget

Compare the global ocean mean net surface heat flux with the time derivative of the ocean heat content (upper 700 m).
The net flux is the sum of the surface net latent, sensible, long-wave and short-wave fluxes minus the latent heat of fusion of the snowfall.

```python
from aqua.diagnostics import OceanHeatBudget, PlotOceanHeatBudget
```

Retrieve the data, compute the two budget terms and save them to NetCDF.

```python
diag = OceanHeatBudget(
    catalog="climatedt-gen2",
    model="IFS-NEMO-5km",
    exp="baseline-hist",
    source="lra-r100-monthly-zarr",
    startdate="1990-01-01",
    enddate="2004-12-31",
)
diag.run(hc_var="hc700m", outputdir="./output")
diag.result
```

Plot the two terms and their difference. The mean of the difference measures how well the budget closes.

```python
figures = PlotOceanHeatBudget(data=diag.result, outputdir="./output").plot(save_format="png", show=True)
```

To plot again from the saved file, without retrieving the data, use `load()` with the same arguments:

```python
diag = OceanHeatBudget(
    catalog="climatedt-gen2",
    model="IFS-NEMO-5km",
    exp="baseline-hist",
    source="lra-r100-monthly-zarr",
    startdate="1990-01-01",
    enddate="2004-12-31",
)
diag.load(hc_var="hc700m", outputdir="./output")
diag.result is not None
```
